```python
import os
from pathlib import Path
from message_source import MessageSource
from event_logger import EventLogger, LoggerStats

log_path = Path("events.log")

source = MessageSource(
    seed=43122,
    total_packets=5000,
    reorder_window=10,
    duplicate_prob=0.05,
    loss_prob=0.02,
    corruption_prob=0.03,
    termination_prob=0.002
)

logger = EventLogger(source, log_path, buffer_size=30)
stats = logger.run()

print("\nLogging completed!")
print(f"Coverage: {stats.packets_written}/{source.total_packets}")
print(f"Inversions: {stats.inversions}")
print(f"Lost in buffer: {stats.final_buffer_size}")
print(f"Corrupted packets: {stats.corrupted_packets}")
print(f"Duplicates discarded: {stats.duplicates_discarded}")
print(f"Retransmits requested: {stats.retransmit_requests}")
print(f"Retransmits received: {stats.retransmits_received}")
print(f"Buffer flushes: {stats.buffer_flushes}")
print(f"Log file path: {log_path.resolve()}")
```
